```python
import pandas as pd
import numpy as np
import os
import ast
```

```python
original_movies_metadata_df = pd.read_csv('..\Data\movies_metadata.csv')
original_movies_metadata_df.head()
```

```python
original_keywords_df = pd.read_csv('..\Data\keywords.csv')
original_keywords_df.head()
```

```python
original_credits_df = pd.read_csv('..\Data\credits.csv')
original_credits_df.head()
```

```python
movies_metadata['id'] = pd.to_numeric(original_movies_metadata_df['id'], errors='coerce')

movies_df = movies_metadata.dropna(subset=['id'])
movies_df['id'] = movies_df['id'].astype('int64')

movies = movies_metadata.merge(original_credits_df, on='id')
movies = movies.merge(original_keywords_df, on='id')

movies.columns
```

```python
movies = movies[['id', 'title', 'genres', 'overview', 'popularity', 'release_date', 'tagline', 'cast', 'crew', 'keywords', 'vote_average', 'vote_count']]
movies.dropna(inplace=True)
movies.head()
```

```python
output_path = r"..\output\normalized_movies_data_part"
os.makedirs(os.path.dirname(output_path), exist_ok=True)
for i, chunk in enumerate(np.array_split(movies, 20)):
    chunk.to_csv(f'{output_path}_{i}.csv', index=False, encoding='utf-8')
```
