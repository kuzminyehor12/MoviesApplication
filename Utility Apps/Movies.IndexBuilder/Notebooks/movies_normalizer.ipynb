```python
import pandas as pd
import numpy as np
import os
import ast
```

```python
original_movies_metadata_df = pd.read_csv('..\Data\movies_metadata.csv')
original_movies_metadata_df.head()
```

```python
original_keywords_df = pd.read_csv('..\Data\keywords.csv')
original_keywords_df.head()
```

```python
original_credits_df = pd.read_csv('..\Data\credits.csv')
original_credits_df.head()
```

```python
original_movies_metadata_df['id'] = pd.to_numeric(original_movies_metadata_df['id'], errors='coerce')

movies_df = original_movies_metadata_df.dropna(subset=['id'])
print(movies_df.shape)

movies_df['id'] = movies_df['id'].astype('int64')

movies = original_movies_metadata_df.merge(original_credits_df, on='id')
movies = movies.merge(original_keywords_df, on='id')

movies.shape
```

```python
movies = movies[['id', 'title', 'genres', 'overview', 'popularity', 'release_date', 'tagline', 'cast', 'crew', 'keywords', 'vote_average', 'vote_count']]
movies['tagline'].fillna('', inplace=True)
movies['popularity'].fillna(0, inplace=True)
movies['vote_average'].fillna(0, inplace=True)
movies['vote_count'].fillna(0, inplace=True)
movies.head(100)
```

```python
def clean_strings(node_string, properties):
    if node_string is None or pd.isna(node_string):
        return None
        
    data = ast.literal_eval(node_string)
    
    for item in data:
        for key in properties:
            if key in item:
                if item[key] is None:
                    item[key] = 'None'
                else:
                    item[key] = item[key].replace("'", "\\'")
                    item[key] = item[key].replace("\"", "")
                    item[key] = item[key].replace("\u00A0", " ")
                    item[key] = item[key].replace("\x92", "\\'")
                    item[key] = item[key].replace("\xad", "-")
    
    return data

movies['keywords'] = movies['keywords'].apply(clean_strings, properties=['name'])
movies['crew'] = movies['crew'].apply(clean_strings, properties=['name', 'job', 'profile_path'])
movies['cast'] = movies['cast'].apply(clean_strings, properties=['name', 'character', 'profile_path'])
```

```python
output_path = r"..\output\normalized_movies_data_part"
os.makedirs(os.path.dirname(output_path), exist_ok=True)
for i, chunk in enumerate(np.array_split(movies, 10)):
    chunk.to_csv(f'{output_path}_{i}.csv', index=False, encoding='utf-8')
```
